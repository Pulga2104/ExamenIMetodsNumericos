```python
import math

a = 1/math.log(10)

def log10 (x, k=5):
    z = float(x)

    for i in range(k):
        z = round(math.sqrt(z),6)
    
    r=z-1
    logz = a*r

    return(2**k) * logz

xi = log10(2)
print(xi)
error_relativo = abs(xi - math.log10(2)) / abs(math.log10(2))
print(error_relativo)
```
