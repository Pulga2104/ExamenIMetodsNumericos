```python
import math

def secant_method_optimized(f, x0, x1, tol=1e-6, max_iter=100):
    x_prev = x0
    x_curr = x1
    iter_count = 0
    
    f_prev = f(x_prev)
    f_curr = f(x_curr)
    
    while abs(f_curr) > tol and iter_count < max_iter:
        x_next = x_curr - f_curr * (x_curr - x_prev) / (f_curr - f_prev)
        
        x_prev = x_curr
        f_prev = f_curr  
        
        x_curr = x_next
        f_curr = f(x_curr) 
        
        iter_count += 1
        
    return x_curr, iter_count


i = 0

def func(x):
    global i
    i += 1
    y = x**3 - 3*x**2 + x - 1
    print(f"Llamada i={i}\t x={x:.5f}\t y={y:.2f}")
    return y



secant_method_optimized(func,x0=2,x1=3)
```

```python
def func1(x):
    global i
    i += 1
    y = x**3 + 2*x**2 + 3*x + 1
    print(f"Llamada i={i}\t x={x:.5f}\t y={y:.2f}")
    return y
    
secant_method_optimized(func1,x0=2,x1=3)
```

```python
def func2(x):
    global i
    i += 1
    y = 35*math.sin(x/math.pi) + 10
    print(f"Llamada i={i}\t x={x:.5f}\t y={y:.2f}")
    return y

secant_method_optimized(func2,x0=2,x1=3)
```
